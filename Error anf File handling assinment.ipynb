### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4 solution code here
with open ('number.txt' , 'w+') as f:
    for i in range(1,6):
        f.write(str(i) + "\n")
    f.seek(0)
    print(f.read())
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
total_sum = 0
with open('number.txt' ,'r') as f:
    for line in f:
        total_sum += int(line.strip())
print(f"Total Sum : {total_sum}")
```

### Q6

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python
# Q8 solution code here
lists = []
with open ("sample.txt" , 'r') as f:
    lines = f.readlines()
    for line in lines:
        lists.append(line.strip())
    print(lists)
```

### Q9

### Q13 (Easy)

### Q17

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
def admin_login(username , password):
    if username == 'admin' and password == 1234:
        return 'Access Granted!!!'
try:
    username = input('Enter your Username')
    password = int(input('Enter your Password'))
    print(admin_login(username , password))
except ValueError:
    print('please enter password in integer')
else:
    print('SuccessFully Login!!')
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def Valid_age (age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Access Granted!! "
try:
    user_age = int(input('Enter your Age'))
    print(Valid_age(user_age))
except ValueError as e :
    print(f'{e}')
```
