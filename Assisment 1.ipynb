### Data Types

```python
integer = 100
float_num = 10.5
string = "hello world"
boolean = True

print(f"Type of integer: {type(integer)} and the value is {integer}")
print(f"Type of float_num: {type(float_num)} and the value is {float_num}")
print(f"Type of string: {type(string)} and the value is {string}")
print(f"Type of boolean: {type(boolean)} and the value is {boolean}")
```

### Arithmetic Operations

```python
Num1= 60
Num2 = 70

print(f"Addition of {Num1} and {Num2} is: {Num1 + Num2}")
print(f"Subtraction of {Num1} and {Num2} is: {Num1 - Num2}")
print(f"Multiplication of {Num1} and {Num2} is: {Num1 * Num2}")
print(f"Division of {Num1} and {Num2} is: {Num1 / Num2}")
print(f"Modulus of {Num1} and {Num2} is: {Num1 % Num2}")
print(f"Power of 2 of {Num1} is: {Num1 ** 2} and {Num2} is: {Num2 ** 2}")
```

###  Rectangle Area

```python
length = 5.0
width = 3.0

area_of_rectangle = length * width
area_of_perimeter = 2 * (length + width)
print("Area of Rectangle:", area_of_rectangle)
print("Area of Perimeter:", area_of_perimeter)
```
